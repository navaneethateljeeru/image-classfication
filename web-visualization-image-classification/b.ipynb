```python
import tensorflow as tf
import cv2
import numpy as np
```

```python
image=cv2.imread("1.jpg")
```

```python
image=image/255.0
```

```python
image=cv2.resize(image,(256,256))
```

```python
classnames=["Audi","Hyundai","mahindra","rolls royce","swift","tata","toyota"]
model=tf.keras.models.load_model("cars.h5")
a=model.predict(np.array([image]))
```

```python
b=np.argmax(a)
```

```python
predict=classnames[b]
```

```python
print("The predicted class is:",predict)
```
